```python

#line chart
df = sns.load_dataset("titanic")
sns.lineplot(data=df,x="pclass", y="fare",hue="sex",marker="*")
#sns.title(lineplot)
print(df)
```

```python
sns.lineplot(data=df,x="age", y="fare",hue = 'sex',marker="*")
plt.title("Scatters")
```

```python
# barchat
df = sns.load_dataset("titanic")
sns.barplot(data=df,x="pclass",y="fare",hue="who")
#sns.title(barplot)
print(df)
```

```python

#sns.scatterplot
df = sns.load_dataset("titanic")
sns.scatterplot(data=df,x="age",y="fare",hue="who")
plt.title(scatter)
print(df)
```

```python
sns.scatterplot(data=df,x="class", y="fare",hue = 'adult_male',marker="*")
plt.title("Scatters")
```

```python
sns.boxplot(data=df,x="age", y="class",hue = 'who')
plt.title("box")
print(df)
```

```python
# seaborn
# heatmap
corr = df[['age','fare','sibsp','parch']].corr()
sns.heatmap(corr,annot=True)
plt.show()
```

```python
# count plot
# count the number of observations in gropup
sns.countplot(data=df,x='pclass',hue='who')
plt.show()
```

```python
 # pairplot ( it represent the scatterplot )
 # multiple numerical values
 # Digonal ( ) and other cells
# different from each numerical column
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.pairplot(df[['age', 'fare', 'sibsp', 'parch']])
print(df)
plt.show()
```

```python
# pivot table + heatmap
# it categorical data to collect
# it is customize
# it is a calculated table where calculate the based on the  conditions
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
pivot = pd.pivot_table(
    df,values='survived',index='sex',columns='class',aggfunc='mean')
print(df)
print(pivot)
```

```python
sns.heatmap(
    pivot,annot=True,fmt='.2f')
```

```python
pivot = pd.pivot_table(
    df,values='survived',index=['sex','class'],aggfunc='mean')
print(df)
print(pivot)
```

```python
sns.heatmap(
    pivot,annot=True,fmt='.2f')
```

```python
# Regression plot - regplot()
# relationship between two numbers
# it is a linear RLSp
# ax+b it is one dimensions
# it is a supervised in a (machine learning languuage)
# it is straight line ( down to up (or ) up to down )
# line up side points plot only data consider points
#Explorantory regression analysis
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.regplot(
    data=df,x='age',y='fare')
print(df)
plt.show()
```

```python
#  DENSITY PLOT (kdeplot())
# cluster ( all apints where that is cluster)
sns.kdeplot(
    data=df,x='age',fill=True)
print(df)
plt.show()
```

```python
sns.kdeplot(
    data=df,x='pclass',fill=True)
print(df)
plt.show()
```

```python
# face grid ( )
g = sns.FacetGrid(
    df,col='age'
)
print(df)
plt.show()
```

```python
# it is only some charts in that grid
# face grid ( )
g = sns.FacetGrid(
    df,col='pclass'
)
g.map(
    sns.histplot,'age'
)
print(df)
plt.show()
```

```python
g = sns.FacetGrid(
    df,col='sex'
)
g.map(
    sns.scatterplot,'age','sex'
)
print(df)
plt.show()
```

```python
# jointplot
sns.jointplot(
    data=df,x='age',y='fare'
)
print(df)
plt.show()
```
